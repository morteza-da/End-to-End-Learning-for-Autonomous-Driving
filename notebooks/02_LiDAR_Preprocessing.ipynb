# Chapter 5 — LiDAR distance encoding (research implementation)

**Source:** PhD research PDF pp. 73–74. This notebook demonstrates the documented planar-distance and RGB encoding formula on **synthetic data only**. The original LiDAR-to-image projection and pixel mapping are unknown and intentionally not invented.

```python
from autonomous_driving.lidar_encoding import planar_distance, distance_to_rgb, encode_points_xy

synthetic_xy = [(0, 0), (6, 8), (0, 20), (0, 30), (0, 40), (0, 50), (0, 60), (0, 70)]
rows = [(xy, planar_distance(*xy), rgb) for xy, rgb in zip(synthetic_xy, encode_points_xy(synthetic_xy))]
for xy, distance, rgb in rows:
    print(f'point={xy}, planar_distance={distance:.1f}, RGB={rgb}')
```

## Color saturation and interpretation

At distance 60 or greater, the red channel saturates at 255 under the printed piecewise rule. The source uses planar distance (Px, Py), not a full 3D Euclidean distance. The conversion below illustrates only **color encoding** and is **not** the full 299×299 LiDAR image creation pipeline.

```python
assert distance_to_rgb(50) == (212, 0, 0) or distance_to_rgb(50) == (213, 0, 0)
assert distance_to_rgb(70) == (255, 0, 0)
print('Synthetic encoding checks passed')
```
